```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import AdaBoostClassifier
from catboost import CatBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

```python
dataset_url = 'https://raw.githubusercontent.com/apogiatzis/breast-cancer-azure-ml-notebook/master/breast-cancer-data.csv'
df = pd.read_csv(dataset_url)
df.head(35)
```

```python
# label encoding
# M = Malignant; B = Benign
df['diagnosis'] = df['diagnosis'].map(
    {
        'M': 1,
        'B': 0
    }
)
```

```python
df.diagnosis.value_counts()
```

```python
features = [
    'radius_mean', 'texture_mean', 'perimeter_mean',
    'smoothness_mean', 'compactness_mean'
]

X=df[features]
y=df.diagnosis
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.19)
```

```python
len(X_train)
```

```python
len(X_test)
```

# **Model Building**

## **Random Forest Model**

```python
model_rf = RandomForestClassifier(
    n_estimators=250, n_jobs=1, random_state=42
)

model_rf.fit(X_train, y_train)
```

```python
model_prediction_rf = model_rf.predict(X_test)
print('Accuracy Score Random Forest Model is: ', round(accuracy_score(model_prediction_rf, y_test)*100,2),'%')
```

## **Support Vector Machine Model**

```python
model_svm = SVC(
    kernel='rbf',
    C=1.0,
    gamma='scale'
)

model_svm.fit(X_train, y_train)
```

```python
model_prediction_svm = model_svm.predict(X_test)
print('Accuracy Score Support Vector Machine model is: ', round(accuracy_score(model_prediction_svm, y_test)*100, 2),'%')
```

## **Decision Tree Model**

```python
model_dt = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)
model_dt.fit(X_train, y_train)
```

```python
model_prediction_dt = model_dt.predict(X_test)
print('Accuracy Score Decision Tree Model is: ', round(accuracy_score(model_prediction_dt, y_test)*100,2),'%')
```

## **AdaBoost Classifier**

```python
model_ab = AdaBoostClassifier(
    estimator=model_dt, n_estimators=100,
    random_state=42
)

model_ab.fit(X_train, y_train)
```

```python
model_prediction_ab = model_ab.predict(X_test)
print('Accuracy Score AdaBoost Model is:', round(accuracy_score(model_prediction_ab, y_test)*100,2),'%')
```

## **CatBoost Model**

```python
model_cb = CatBoostClassifier(
    iterations=400, learning_rate=0.1, depth=3
)

model_cb.fit(X_train, y_train)
```

```python
model_prediction_cb = model_cb.predict(X_test)
print('Accuracy Score CatBoost Model is:', round(accuracy_score(model_prediction_cb, y_test)*100,2),'%')
```
